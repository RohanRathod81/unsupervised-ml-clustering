```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris =load_iris()

X =iris.data
y=iris.target
```

```python
# Scaling the X values

scaler =StandardScaler()
X_scaled =scaler.fit_transform(X)
```

```python
# Visualize 

sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:,2])
```

```python
# Import dendrogram from scipy module 

from scipy.cluster.hierarchy import linkage, dendrogram

# Linkage matrics
Z = linkage(X_scaled, method ="ward")


# plot Dendogram
plt.figure(figsize=(12,6))
dendrogram(Z)
plt.xlabel("samples")
plt.ylabel("distance")
plt.title("Dendogram for hierarchical clustering")
```

```python
# Clustering 
from sklearn.cluster import AgglomerativeClustering

agg = AgglomerativeClustering(n_clusters=3)
lables = agg.fit_predict(X_scaled)
```

```python
# Visualize using c=labels

sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:,2], c=lables)
```
